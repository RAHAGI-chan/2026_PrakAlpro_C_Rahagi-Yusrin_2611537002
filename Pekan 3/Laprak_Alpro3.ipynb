## 2.1 Dasar Teori

Operator Aritmatika

Digunakan untuk melakukan perhitungan matematis.

Contoh:

+ (penjumlahan)

- (pengurangan)

* (perkalian)

/ (pembagian)

// (pembagian bulat)

% (sisa bagi/modulus)

** (pangkat)

Membantu dalam pengolahan data numerik seperti menghitung total belanja, rata-rata, atau sisa pembagian.

Operator Assignment (Penugasan)

Digunakan untuk memberikan nilai ke variabel atau memperbarui nilai variabel dengan operasi tertentu.

Contoh:

= (assignment biasa)

+=, -=, *=, /=, //=, %= (augmented assignment)

**= (assignment pangkat)

Berguna untuk memperbarui nilai variabel secara langsung tanpa harus menulis ulang ekspresi.

Operator Perbandingan

Digunakan untuk membandingkan dua nilai dan menghasilkan output boolean (True atau False).

Contoh:

== (sama dengan)

!= (tidak sama dengan)

> (lebih besar)

< (lebih kecil)

>= (lebih besar atau sama dengan)

<= (lebih kecil atau sama dengan)

Membantu dalam validasi syarat, misalnya apakah belanja ≥ Rp200.000.

Operator Logika

Digunakan untuk menggabungkan kondisi boolean.

Contoh:

and → True jika kedua kondisi benar.

or → True jika salah satu kondisi benar.

not → membalik nilai boolean.

Berguna dalam pengambilan keputusan kompleks, misalnya promo hanya berlaku jika pelanggan member dan belanja ≥ Rp200.000.

Operator Keanggotaan

Digunakan untuk memeriksa apakah suatu elemen ada dalam koleksi (list, string, dll).

Contoh:

in → True jika elemen ada.

not in → True jika elemen tidak ada.

Berguna untuk validasi kode promo dalam daftar promo.

Operator Identitas

Digunakan untuk membandingkan identitas objek (apakah dua variabel menunjuk ke objek yang sama di memori).

Contoh:

is → True jika identitas objek sama.

is not → True jika identitas objek berbeda.

Berbeda dengan == yang membandingkan nilai/isi objek.

Operator Bitwise

Digunakan untuk operasi pada level bit biner.

Contoh:

& (AND)

| (OR)

^ (XOR)

~ (NOT)

<< (geser kiri)

>> (geser kanan)

Berguna untuk representasi status dalam bentuk kode biner, misalnya hak akses pelanggan.

### 2.2.1 Program "Kelas C_2611537002"


```python
# Buat program untuk operator aritmatika dalam python
# Nama variabel ditambah 4 digit NIM terakhir, contoh: angka1_1234
# Progam ini menggunakan fungsi input ()
# Nilai yang dimasukkan akan dikonveri menjadi tipe data integer

angka1_7002 = int(input("Input angka-1: "))
angka2_7002 = int(input("Input angka-2: "))

# Penjumlahan
hasil = angka1_7002 + angka2_7002
print("\nOperator Penjumlahan")
print("hasil =", hasil)


# Perkalian
hasil = angka1_7002 * angka2_7002
print("\nOperator Perkalian")
print("Hasil =", hasil)

# Pembagian, Pembagian Bulat, dan sisa bagi
if angka2_7002 != 0:
    hasil = angka1_7002 / angka2_7002
    print("\nOperator Pembagian")
    print("Hasil =", hasil)

    hasil = angka1_7002 // angka2_7002
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil)

    hasil = angka1_7002 % angka2_7002
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil)
else:
    print("angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil = angka1_7002** angka2_7002
print("\nOperator Pangkat")
print("Hasil =", hasil)
```

Menunjukkan bagaimana Python melakukan operasi dasar matematika, termasuk pengecekan pembagian dengan nol.

```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1 = int(input("Input angka-1: "))
angka2 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1)
print("Nilai angka2 =", angka2)

# Assignment biasa
hasil = angka1
print("\nAssignment biasa (=)")
print("Hasil =", hasil)

# Assignment penambahan
hasil = angka1
hasil += angka2
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil)

# Assignment pengurangan
hasil = angka1
hasil -= angka2
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil)

# Assignment perkalian
hasil = angka1
hasil *= angka2
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil)

# Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2 != 0:
    hasil = angka1
    hasil /= angka2
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil)
    # Operator tambahan
    hasil = angka1
    hasil //= angka2
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil)
    hasil = angka1
    hasil %= angka2
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil = angka1
hasil **= angka2
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil)
```

Program Keanggotaan dan Identitas

Identitas (is, is not) biasanya digunakan untuk membandingkan apakah dua variabel menunjuk ke objek yang sama.


```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menggunakan fungsi input()
# Program operator logika dalam Python

# Memasukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_7002 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_7002 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 =", a1_7002)
print("A2 =", a2_7002)

# Konjungsi: bernilai True jika keduanya True
hasil = a1_7002 and a2_7002
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil)

# Disjungsi: bernilai True jika salah satunya True
hasil = a1_7002 or a2_7002
print("\nDisjungsi (OR)")
print("A1 or A2 =", hasil)

# Negasi A1: membalik nilai A1
hasil = not a1_7002
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil)

# Negasi A2: membalik nilai A2
hasil = not a2_7002
print("\nNegasi A2 (NOT)")
print("not A2 =", hasil)

# XOR: bernilai True jika kedua nilai berbeda
hasil = a1_7002 != a2_7002
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil)
```
