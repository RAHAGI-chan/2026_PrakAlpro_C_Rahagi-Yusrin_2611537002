Nested adalah penggunaan "for" yang didalamnya terdapat "for" yang lainnya sehingga memberikan variabel (lebih dari satu) aksi tertentu dan keberadaan serta jarak pada variabel tersebut.




## 2.2 Kode Program

### 2.2.1 Program "Kelas C_2611537002"


Kode dibawah ini adalah contoh dari penerapan metode perulangan didalam python.

```python
ulang_7002 = int(input("Masukkan nilai batas: "))

jumlah_7002 = 0
for i_7002 in range(1, ulang_7002 + 1):
    if i_7002 % 2 == 0:
        print(i_7002, end=" ")
        jumlah_7002 = jumlah_7002 + i_7002
    if i_7002 < ulang_7002:
        print(" = ", end="")
    else:
        print(" = ", jumlah_7002, end="")
print()
print("jumlah = ", jumlah_7002)
```

```python
batas_7002 = int(input("Masukkan nilai batas: "))
for line in range(1, batas_7002 + 1):
    for j_7002 in range(1, (-1 * line + batas_7002) + 1):
        print(".", end="")
print(line)
```

**Penjelasan :**

```python
batas_7002 = int(input("Masukkan nilai batas: "))
for i_7002 in range(1, batas_7002 + 1):
    for j_7002 in range(1, batas_7002 + 1):
        print("*", end="")
print() 
```

Program diatas dapat melakukan perulangan dalam bentuk kolom dan mengulang angka yang diinput oleh user dua kali secara berurutan.

```python
tinggi_7002 = int(input("Masukkan tinggi pola (Bilangan genap, misal 10): "))

if tinggi_7002 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_7002 = tinggi_7002
    c_7002 = a_7002
    lebar_7002 = (2 * tinggi_7002) - 2

    for i_7002 in range(1, tinggi_7002 +1):
        b_7002 = c_7002 + 1

        for j_7002 in range(1, lebar_7002 + 1):

            # Baris atas dan bawah
            if i_7002 == 1 or i_7002 == tinggi_7002:
                if j_7002 == 1 or j_7002 == lebar_7002:
                    print("#", end="")
                else:
                    print("=", end="")
            
            # Baris isi

            else:
                if j_7002 == 1 or j_7002 == lebar_7002:
                    print("|", end="")
                else:
                    if j_7002 == c_7002:
                        print("<", end="")
                    elif j_7002 == b_7002:
                        print(">", end="")
                    elif j_7002 == (lebar_7002 - c_7002 - 1):
                        print("<", end="")
                    elif j_7002 == (lebar_7002 - c_7002 + 1):
                        print(">", end="")
                    elif j_7002> b_7002 and j_7002 < (lebar_7002 - c_7002):
                        print(",", end="")
                    else:
                        print (" ", end="")
        print()

        # Logika asli j_7002ava
        a_7002 -= 2

        if a_7002 <= 0:
            c_7002 = (-a_7002) + 2
        else:
            c_7002 = a_7002
```

Program diatas dapat membuat suatu pola yang dirancang menggunakan kombinasi metode perulangan dengan memasukkan tinggi dan membuat lebar dari tinggi itu sendiri yang diinputkan oleh user.

```python
# Buat program untuk perulangan for dalam python
# Program ini menggunakan fungsi input()

ulang_7002 = int(input("Masukkan jumlah perulangan: "))

for i_7002 in range(ulang_7002):
    print(f"Perulangan ke-{i_7002}")
```

Program diatas adalah program yang dapat melakukan penyebutan suatu angka yang diinput oleh user secara berulang dalam bentuk kolom.

```python
ulang_7002 = int(input("Masukkan jumlah perulangan: "))
print("Perulangan ke-0 sampai ke-", ulang_7002-1)
for i_7002 in range(ulang_7002):
    print(i_7002, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_7002)
for i_7002 in range(1, ulang_7002 + 1):
    print(i_7002, end=" ")
```

Program diatas dapat melakukan perulangan dengan memasukkan angka yang diinginkan oleh user dan program akan mengulang angka tersebut menyamping dan tidak dalam bentuk kolom.

```python
ulang_7002 = int(input("Masukkan jumlah perulangan: "))

jumlah_7002 = 0
for i_7002 in range (1, ulang_7002 + 1):
    print(i_7002, end=" ")
    jumlah_7002 = jumlah_7002 + i_7002

    if i_7002 < ulang_7002:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_7002, end=" ")
print()
print("jumlah =", jumlah_7002)
```

Program diatas dapat melakukan penjumlahan deretan dengan memasukkan angka yang ingin di eksekusi.

```python
tinggi_7002 = int(input("Masukkan nilai segitiga: "))

for i_7002 in range(1, tinggi_7002 + 1):
    print("", end=" ")

    for j_7002 in range(tinggi_7002 - i_7002):
        print("", end=" ")
          
    for j_7002 in range(i_7002):
        print("*", end=" ")

    print()
```

Program diatas berfungsi untuk membuat pola segitiga dengan memberikan angka yang dapat diproses menjadi bentuk segitiga.
