## 2.1 Dasar Teori
Konsep utama: if digunakan untuk membuat percabangan logika, yaitu memilih jalur eksekusi program berdasarkan kondisi tertentu.

Evaluasi kondisi: Kondisi yang ditulis dalam if harus menghasilkan nilai boolean, yaitu True atau False.

Indikator kebenaran (truthiness): Nilai kosong seperti 0, string kosong, list kosong, dictionary kosong, dan None dianggap False. Nilai selain itu dianggap True.

Indentasi: Python menggunakan indentasi (spasi/tab) untuk menentukan blok kode yang dijalankan. Tanpa indentasi yang konsisten, program akan error.

Variasi penggunaan:

if tunggal → hanya mengeksekusi blok kode jika kondisi benar.

if-else → menangani dua kemungkinan, kondisi benar atau salah.

if-elif-else → menangani lebih dari dua kondisi secara berurutan.

nested if → percabangan bersarang untuk logika yang lebih kompleks.

inline if (ternary operator) → bentuk singkat untuk memilih nilai berdasarkan kondisi.

Operator logika: and, or, dan not digunakan untuk menggabungkan atau memodifikasi kondisi.

Urutan kondisi: Harus logis, dari yang paling spesifik ke umum, agar hasil evaluasi tidak salah.

Peran dalam program: if adalah dasar kontrol alur program. Tanpa percabangan, program hanya berjalan linear tanpa kemampuan menyesuaikan diri terhadap data atau situasi.

Kegunaan praktis: Digunakan dalam validasi input, pengambilan keputusan, kontrol menu, sistem penilaian, dan berbagai logika aplikasi.

Program tersebut adalah program yang dapat digunakan untuk kasir dan bisa memasukkan potongan diskon terhadap nominal harga yang ditetapkan sehingga pembeli yang memiliki diskon mendapatkan potongan harga dengan menjadi member pada perusahaan tersebut.
